```python
import os
import time
import shutil
from tqdm import tqdm
from docling.document_converter import DocumentConverter

def prepare_output_directories(directories: list[str]) -> None:
    """
    Removes and recreates output directories for converted Markdown files.

    Args:
        directories (list[str]): List of directory names to prepare.
    """
    for directory in directories:
        if os.path.exists(directory):
            shutil.rmtree(directory)
        os.makedirs(directory, exist_ok=True)

def convert_pdfs(pdf_dir: str, output_dir: str, do_ocr: bool = False, backend: str = None, limit: int = None) -> float:
    """
    Converts PDF files in a directory to Markdown using Docling.

    Args:
        pdf_dir (str): Directory containing PDF files.
        output_dir (str): Directory to save the Markdown files.
        do_ocr (bool): Whether to use OCR during conversion.
        backend (str): Backend option for DocumentConverter ('v2', etc.).
        limit (int): Maximum number of PDFs to process.

    Returns:
        float: Time taken for the conversion process, in seconds.
    """
    pdf_files = [f for f in os.listdir(pdf_dir) if f.endswith('.pdf')]
    if limit:
        pdf_files = pdf_files[:limit]

    start_time = time.time()

    for pdf_file in tqdm(pdf_files, desc=f"Converting PDFs to {output_dir}"):
        try:
            pdf_path = os.path.join(pdf_dir, pdf_file)
            md_filename = pdf_file.replace('.pdf', '.md')
            md_path = os.path.join(output_dir, md_filename)

            if os.path.exists(md_path):
                continue

            converter = DocumentConverter()
            converter.do_ocr = do_ocr
            if backend:
                converter.backend = backend

            doc = converter.convert(pdf_path)

            with open(md_path, 'w', encoding='utf-8') as f:
                f.write(doc.document.export_to_markdown())

        except Exception as e:
            print(f"Error converting {pdf_file}: {str(e)}")

    end_time = time.time()
    return end_time - start_time

def test_docling_conversion(pdf_dir: str = 'charakterystyki_pdf', limit: int = 1) -> None:
    """
    Tests Docling document conversion with OCR, without OCR, and using backend v2.

    Args:
        pdf_dir (str): Directory containing the input PDFs.
        limit (int): Number of PDFs to process for the test.
    """
    output_dirs = ['charakterystyki_md_ocr', 'charakterystyki_md_nocr', 'charakterystyki_md_v2']
    prepare_output_directories(output_dirs)

    print("Starting Docling conversion tests...")

    # Conversion with OCR
    time_ocr = convert_pdfs(pdf_dir, 'charakterystyki_md_ocr', do_ocr=True, limit=limit)
    print(f"Conversion with OCR completed in {time_ocr:.2f} seconds.")

    # Conversion without OCR
    time_nocr = convert_pdfs(pdf_dir, 'charakterystyki_md_nocr', do_ocr=False, limit=limit)
    print(f"Conversion without OCR completed in {time_nocr:.2f} seconds.")

    # Conversion with backend 'v2'
    time_v2 = convert_pdfs(pdf_dir, 'charakterystyki_md_v2', backend='v2', limit=limit)
    print(f"Conversion with backend 'v2' completed in {time_v2:.2f} seconds.")

# --- Usage Example ---
test_docling_conversion(limit=1)
```

```python
import os
import time
import shutil
from pathlib import Path
from tqdm import tqdm
from docling.document_converter import DocumentConverter
from docling.datamodel.base_models import ConversionStatus

def convert_pdfs_loop(pdf_dir: str, output_dir: str, pdf_files: list[str]) -> float:
    """
    Converts PDFs to Markdown individually in a loop using Docling.

    Args:
        pdf_dir (str): Directory containing PDF files.
        output_dir (str): Directory to save Markdown files.
        pdf_files (list[str]): List of PDF filenames to convert.

    Returns:
        float: Time taken for the looped conversion process, in seconds.
    """
    start_time = time.time()

    for pdf_file in tqdm(pdf_files, desc=f"Loop Conversion to {output_dir}"):
        try:
            pdf_path = os.path.join(pdf_dir, pdf_file)
            md_filename = pdf_file.replace('.pdf', '.md')
            md_path = os.path.join(output_dir, md_filename)

            if os.path.exists(md_path):
                continue

            converter = DocumentConverter()
            doc = converter.convert(pdf_path)

            with open(md_path, 'w', encoding='utf-8') as f:
                f.write(doc.document.export_to_markdown())

        except Exception as e:
            print(f"Error converting {pdf_file}: {str(e)}")

    end_time = time.time()
    return end_time - start_time

def convert_pdfs_batch(pdf_dir: str, output_dir: str, pdf_files: list[str]) -> float:
    """
    Converts PDFs to Markdown using Docling's batch method.

    Args:
        pdf_dir (str): Directory containing PDF files.
        output_dir (str): Directory to save Markdown files.
        pdf_files (list[str]): List of PDF filenames to convert.

    Returns:
        float: Time taken for the batch conversion process, in seconds.
    """
    pdf_paths = [os.path.join(pdf_dir, pdf_file) for pdf_file in pdf_files]

    start_time = time.time()

    converter = DocumentConverter()
    conv_results = converter.convert_all(
        pdf_paths,
        raises_on_error=False,
    )

    for result in conv_results:
        md_path = Path(output_dir)
        doc_filename = result.input.file.stem

        if result.status == ConversionStatus.SUCCESS:
            # Save the result as Markdown
            with (md_path / f"{doc_filename}.md").open("w", encoding="utf-8") as f:
                f.write(result.document.export_to_markdown())
        else:
            print(f"Conversion failed for {doc_filename}: {result.error}")

    end_time = time.time()
    return end_time - start_time

def test_docling_loop_vs_batch(pdf_dir: str = 'charakterystyki_pdf', limit: int = 20) -> None:
    """
    Tests Docling PDF conversion: loop-based vs batch-based methods.

    Args:
        pdf_dir (str): Directory containing the input PDFs.
        limit (int): Number of PDFs to process for the test.
    """
    output_dirs = ['charakterystyki_md_loop', 'charakterystyki_md_batch']
    prepare_output_directories(output_dirs)

    print("Starting Docling conversion test: loop vs batch...")

    # Collect PDF files
    pdf_files = [f for f in os.listdir(pdf_dir) if f.endswith('.pdf')]
    pdf_files = pdf_files[:limit]

    # Loop-based conversion
    time_loop = convert_pdfs_loop(pdf_dir, 'charakterystyki_md_loop', pdf_files)
    print(f"Loop conversion completed in {time_loop:.2f} seconds.")

    # Batch-based conversion
    time_batch = convert_pdfs_batch(pdf_dir, 'charakterystyki_md_batch', pdf_files)
    print(f"Batch conversion completed in {time_batch:.2f} seconds.")

test_docling_loop_vs_batch(limit=20)
```

```python
from google import genai
from google.genai import types
import os
from tqdm import tqdm
import requests
import time
import json

# Konfiguracja API Gemini
# Wczytanie klucza API z pliku secrets.json
with open('secrets.json') as f:
    secrets = json.loads(f.read())
    GOOGLE_API_KEY = secrets['GOOGLE_API_KEY']
client = genai.Client(api_key=GOOGLE_API_KEY)

# Utworzenie katalogu na pliki MD, jeśli nie istnieje
if not os.path.exists('charakterystyki_md_gemini'):
    os.makedirs('charakterystyki_md_gemini')

print("Rozpoczynam konwersję plików PDF na MD używając Gemini...")

# Pobranie linków z DataFrame
# pdf_links = df_char['Charakterystyka'].dropna().tolist()
# Pobranie pierwszych 90 linków z DataFrame
pdf_links = df_char['Charakterystyka'].dropna().head(90).tolist()


# Funkcja do pobierania zawartości PDF
def pobierz_pdf(url):
    try:
        response = requests.get(url, timeout=30)
        if response.status_code == 200:
            return response.content
        return None
    except Exception as e:
        print(f"Błąd podczas pobierania PDF: {str(e)}")
        return None

# Funkcja do konwersji PDF na MD używając Gemini
def konwertuj_na_md(pdf_content):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=[
                types.Part.from_bytes(
                    data=pdf_content,
                    mime_type='application/pdf',
                ),
                "Przekonwertuj ten dokument PDF na format Markdown zachowując strukturę i formatowanie."
            ]
        )
        return response.text
    except Exception as e:
        print(f"Błąd podczas konwersji: {str(e)}")
        return None

# Przetwarzanie każdego linku
for i, link in enumerate(tqdm(pdf_links)):
    try:
        # Nazwa pliku MD
        md_filename = f"charakterystyka_{i}.md"
        md_path = os.path.join('charakterystyki_md_gemini', md_filename)
        
        # Pobranie PDF
        pdf_content = pobierz_pdf(link)
        if pdf_content is None:
            continue
            
        # Konwersja na MD
        md_content = konwertuj_na_md(pdf_content)
        if md_content is None:
            continue
            
        # Zapisanie pliku MD
        with open(md_path, 'w', encoding='utf-8') as f:
            f.write(md_content)
            
        # Opóźnienie między requestami aby uniknąć limitów API
        time.sleep(2)
        
    except Exception as e:
        print(f"Wystąpił błąd przy przetwarzaniu linku {link}: {str(e)}")
        continue

print("\nZakończono konwersję plików PDF na MD")
print(f"Zapisano pliki w katalogu 'charakterystyki_md_gemini'")
```

```python
from google import genai
from google.genai import types
import os
from tqdm import tqdm
import requests
import pathlib
import time
import json

# Konfiguracja API Gemini
# Wczytanie klucza API z pliku secrets.json
with open('secrets.json') as f:
    secrets = json.loads(f.read())
    GOOGLE_API_KEY = secrets['GOOGLE_API_KEY']
client = genai.Client(api_key=GOOGLE_API_KEY)

# Utworzenie katalogu na pliki MD, jeśli nie istnieje
if not os.path.exists('charakterystyki_md_gemini_lokalnie'):
    os.makedirs('charakterystyki_md_gemini_lokalnie')

print("Rozpoczynam konwersję plików PDF na MD używając Gemini...")

# Pobranie ścieżek do lokalnych plików PDF
pdf_paths = [f"charakterystyki_pdf/charakterystyka_{i}.pdf" for i in range(90)]

# Funkcja do konwersji PDF na MD używając Gemini
def konwertuj_na_md(pdf_content):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=[
                types.Part.from_bytes(
                    data=pdf_content,
                    mime_type='application/pdf',
                ),
                "Przekonwertuj ten dokument PDF na format Markdown zachowując strukturę i formatowanie."
            ]
        )
        return response.text
    except Exception as e:
        print(f"Błąd podczas konwersji: {str(e)}")
        return None

# Przetwarzanie każdego linku
for i, path in enumerate(tqdm(pdf_paths)):
    try:
        # Nazwa pliku MD
        md_filename = f"charakterystyka_{i}.md"
        md_path = os.path.join('charakterystyki_md_gemini_lokalnie', md_filename)
        
        if not os.path.exists(path):
            print(f"Plik {path} nie istnieje")
            continue
            
        # Odczytanie zawartości pliku PDF
        with open(path, 'rb') as pdf_file:
            pdf_content = pdf_file.read()
            
        # Konwersja na MD
        md_content = konwertuj_na_md(pdf_content)
        if md_content is None:
            continue
            
        # Zapisanie pliku MD
        with open(md_path, 'w', encoding='utf-8') as f:
            f.write(md_content)
            
        # Opóźnienie między requestami aby uniknąć limitów API
        time.sleep(2)
        
    except Exception as e:
        print(f"Wystąpił błąd przy przetwarzaniu pliku {path}: {str(e)}")
        continue

print("\nZakończono konwersję plików PDF na MD")
print(f"Zapisano pliki w katalogu 'charakterystyki_md_gemini'")
```
