```python
import requests

url = "https://rejestry.ezdrowie.gov.pl/api/rpl/medicinal-products/public-pl-report/get-csv"
response = requests.get(url)

with open('Rejestr_Produktow_Leczniczych.csv', 'wb') as f:
    f.write(response.content)
```

```python
import pandas as pd

df = pd.read_csv('Rejestr_Produktow_Leczniczych.csv', sep=';', quotechar='"')
```
