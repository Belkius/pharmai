```python
# Tworzenie nowego dataframe zawierającego tylko identyfikator i charakterystykę
df_char = df[['Identyfikator Produktu Leczniczego', 'Charakterystyka']]

# Usuwanie wierszy z pustą charakterystyką
df_char = df_char.dropna(subset=['Charakterystyka'])

# Wyświetlenie pierwszych 5 wierszy nowego dataframe
print("Pierwsze 5 wierszy dataframe z identyfikatorami i charakterystykami:")
print(df_char.head())

# Podstawowe informacje o nowym dataframe
print("\nInformacje o dataframe:")
print(f"Liczba wierszy: {len(df_char)}")
print(f"Liczba wierszy z niepustą charakterystyką: {df_char['Charakterystyka'].notna().sum()}")
```

```python
import requests
import time
import os
from tqdm import tqdm

# Funkcja do pobierania zawartości PDF
def pobierz_pdf(url):
    try:
        response = requests.get(url, timeout=30)
        if response.status_code == 200:
            return response.content
        return None
    except Exception as e:
        print(f"Błąd podczas pobierania PDF: {str(e)}")
        return None
    
def zapisz_pdf(pdf_content, filename):
    with open(filename, 'wb') as f:
        f.write(pdf_content)

# Utworzenie katalogu na pliki PDF, jeśli nie istnieje
if not os.path.exists('charakterystyki_pdf'):
    os.makedirs('charakterystyki_pdf')

# Pobranie pierwszych linków do PDF
pdf_links = df_char['Charakterystyka']
# Pobranie 90 linków
pdf_links = pdf_links.head(90)
print("Rozpoczynam pobieranie plików PDF...")

# Iteracja po linkach
for i, link in enumerate(tqdm(pdf_links)):
    try:
        # Pobranie pliku PDF
        pdf_content = pobierz_pdf(link)
        if pdf_content is None:
            continue
        
        # Zapisanie pliku PDF
        filename = f'charakterystyki_pdf/charakterystyka_{i+1}.pdf'
        zapisz_pdf(pdf_content, filename)
        # print(f"Plik PDF zapisany jako {filename}")
                    
        # Opóźnienie między requestami aby uniknąć limitów API
        time.sleep(2)

    except Exception as e:
        print(f"Wystąpił błąd przy pobieraniu pliku {i+1}: {str(e)}")

print("\nZakończono pobieranie plików PDF")
print(f"Zapisano pliki w katalogu 'charakterystyki_pdf'")
```

```python
from docling.document_converter import DocumentConverter
import os
from tqdm import tqdm

# Utworzenie katalogu na pliki MD, jeśli nie istnieje
if not os.path.exists('charakterystyki_md'):
    os.makedirs('charakterystyki_md')

print("Rozpoczynam konwersję plików PDF na MD...")

# Ścieżka do katalogu z plikami PDF
pdf_dir = 'charakterystyki_pdf'

# Lista wszystkich plików PDF w katalogu
pdf_files = [f for f in os.listdir(pdf_dir) if f.endswith('.pdf')]

# Konwersja każdego pliku PDF na MD
for pdf_file in tqdm(pdf_files):
    try:
        # Pełna ścieżka do pliku PDF
        pdf_path = os.path.join(pdf_dir, pdf_file)
        
        # Nazwa pliku MD (zamiana rozszerzenia .pdf na .md)
        md_filename = pdf_file.replace('.pdf', '.md')
        md_path = os.path.join('charakterystyki_md', md_filename)
        
        converter = DocumentConverter()

        # Konwersja PDF na MD używając docling
        doc = converter.convert(pdf_path)
        md_content = doc.document.export_to_markdown()
        
        # Zapisanie pliku MD
        with open(md_path, 'w', encoding='utf-8') as f:
            f.write(md_content)
            
    except Exception as e:
        print(f"Wystąpił błąd przy konwersji pliku {pdf_file}: {str(e)}")

print("\nZakończono konwersję plików PDF na MD")
print(f"Zapisano pliki w katalogu 'charakterystyki_md'")
```

```python
from google import genai
from google.genai import types
import os
from tqdm import tqdm
import requests
import time
import json

# Konfiguracja API Gemini
# Wczytanie klucza API z pliku secrets.json
with open('secrets.json') as f:
    secrets = json.loads(f.read())
    GOOGLE_API_KEY = secrets['GOOGLE_API_KEY']
client = genai.Client(api_key=GOOGLE_API_KEY)

# Utworzenie katalogu na pliki MD, jeśli nie istnieje
if not os.path.exists('charakterystyki_md_gemini'):
    os.makedirs('charakterystyki_md_gemini')

print("Rozpoczynam konwersję plików PDF na MD używając Gemini...")

# Pobranie linków z DataFrame
# pdf_links = df_char['Charakterystyka'].dropna().tolist()
# Pobranie pierwszych 90 linków z DataFrame
pdf_links = df_char['Charakterystyka'].dropna().head(90).tolist()


# Funkcja do pobierania zawartości PDF
def pobierz_pdf(url):
    try:
        response = requests.get(url, timeout=30)
        if response.status_code == 200:
            return response.content
        return None
    except Exception as e:
        print(f"Błąd podczas pobierania PDF: {str(e)}")
        return None

# Funkcja do konwersji PDF na MD używając Gemini
def konwertuj_na_md(pdf_content):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=[
                types.Part.from_bytes(
                    data=pdf_content,
                    mime_type='application/pdf',
                ),
                "Przekonwertuj ten dokument PDF na format Markdown zachowując strukturę i formatowanie."
            ]
        )
        return response.text
    except Exception as e:
        print(f"Błąd podczas konwersji: {str(e)}")
        return None

# Przetwarzanie każdego linku
for i, link in enumerate(tqdm(pdf_links)):
    try:
        # Nazwa pliku MD
        md_filename = f"charakterystyka_{i}.md"
        md_path = os.path.join('charakterystyki_md_gemini', md_filename)
        
        # Pobranie PDF
        pdf_content = pobierz_pdf(link)
        if pdf_content is None:
            continue
            
        # Konwersja na MD
        md_content = konwertuj_na_md(pdf_content)
        if md_content is None:
            continue
            
        # Zapisanie pliku MD
        with open(md_path, 'w', encoding='utf-8') as f:
            f.write(md_content)
            
        # Opóźnienie między requestami aby uniknąć limitów API
        time.sleep(2)
        
    except Exception as e:
        print(f"Wystąpił błąd przy przetwarzaniu linku {link}: {str(e)}")
        continue

print("\nZakończono konwersję plików PDF na MD")
print(f"Zapisano pliki w katalogu 'charakterystyki_md_gemini'")
```

```python
from google import genai
from google.genai import types
import os
from tqdm import tqdm
import requests
import pathlib
import time
import json

# Konfiguracja API Gemini
# Wczytanie klucza API z pliku secrets.json
with open('secrets.json') as f:
    secrets = json.loads(f.read())
    GOOGLE_API_KEY = secrets['GOOGLE_API_KEY']
client = genai.Client(api_key=GOOGLE_API_KEY)

# Utworzenie katalogu na pliki MD, jeśli nie istnieje
if not os.path.exists('charakterystyki_md_gemini_lokalnie'):
    os.makedirs('charakterystyki_md_gemini_lokalnie')

print("Rozpoczynam konwersję plików PDF na MD używając Gemini...")

# Pobranie ścieżek do lokalnych plików PDF
pdf_paths = [f"charakterystyki_pdf/charakterystyka_{i}.pdf" for i in range(90)]

# Funkcja do konwersji PDF na MD używając Gemini
def konwertuj_na_md(pdf_content):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=[
                types.Part.from_bytes(
                    data=pdf_content,
                    mime_type='application/pdf',
                ),
                "Przekonwertuj ten dokument PDF na format Markdown zachowując strukturę i formatowanie."
            ]
        )
        return response.text
    except Exception as e:
        print(f"Błąd podczas konwersji: {str(e)}")
        return None

# Przetwarzanie każdego linku
for i, path in enumerate(tqdm(pdf_paths)):
    try:
        # Nazwa pliku MD
        md_filename = f"charakterystyka_{i}.md"
        md_path = os.path.join('charakterystyki_md_gemini_lokalnie', md_filename)
        
        if not os.path.exists(path):
            print(f"Plik {path} nie istnieje")
            continue
            
        # Odczytanie zawartości pliku PDF
        with open(path, 'rb') as pdf_file:
            pdf_content = pdf_file.read()
            
        # Konwersja na MD
        md_content = konwertuj_na_md(pdf_content)
        if md_content is None:
            continue
            
        # Zapisanie pliku MD
        with open(md_path, 'w', encoding='utf-8') as f:
            f.write(md_content)
            
        # Opóźnienie między requestami aby uniknąć limitów API
        time.sleep(2)
        
    except Exception as e:
        print(f"Wystąpił błąd przy przetwarzaniu pliku {path}: {str(e)}")
        continue

print("\nZakończono konwersję plików PDF na MD")
print(f"Zapisano pliki w katalogu 'charakterystyki_md_gemini'")
```
