```python
import matplotlib.pyplot as plt
import os

avg_by_question = df.groupby(["Pytanie", "System"]).agg({
    "Trafność": "mean",
    "Precyzja": "mean"
}).reset_index()

systemy = avg_by_question["System"].unique()

os.makedirs("wykresy", exist_ok=True)

for system in systemy:
    system_df = avg_by_question[avg_by_question["System"] == system]

    plt.figure(figsize=(10, 5))
    plt.plot(system_df["Pytanie"], system_df["Trafność"], marker='o', label='Trafność', color="#a6cee3")
    plt.plot(system_df["Pytanie"], system_df["Precyzja"], marker='s', label='Precyzja', color="#fdbf6f")

    plt.title(f"Średnie oceny trafności i precyzji dla {system}")
    plt.xlabel("Numer pytania")
    plt.ylabel("Średnia ocena dla pytania")
    plt.xticks(ticks=range(1, 22))
    plt.ylim(0, 5.5)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.tight_layout()

    filename = f"wykresy/wykres_{system.replace(' ', '_')}.png"
    plt.savefig(filename, dpi=300)
    plt.close()
```
